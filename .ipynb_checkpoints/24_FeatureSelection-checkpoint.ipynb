```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
IRIS = pd.read_csv("Datasets/IRIS.csv")
IRIS
```

```python
# Shape and summary
print(IRIS.info())
print(IRIS.shape)
print(IRIS.describe())
print(IRIS['species'].value_counts())
```

```python
# Check missing and duplicates
print(IRIS.isnull().sum())
print("Duplicates:", IRIS.duplicated().sum())
IRIS.nunique()
```

```python
sepal_length = IRIS.iloc[:, 0] # Extract sepal length (feature at index 0)

mu, std = np.mean(sepal_length), np.std(sepal_length)
x = np.linspace(np.min(sepal_length), np.max(sepal_length), 100)
y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / std)**2)

plt.figure(figsize=(8, 6))
plt.hist(sepal_length, bins=20, color='skyblue', edgecolor='black', alpha=0.7, density=True)
plt.plot(x, y, color='red', label='Gaussian Fit')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Density')
plt.title('Distribution of Sepal Length in Iris Dataset with Gaussian Fit')
plt.legend()
plt.show()
```

```python
np.linspace(np.min(sepal_length), np.max(sepal_length), 100)
```

```python
sepal_width = IRIS.iloc[:, 1] # Extract sepal width (feature at index 0)

mu, std = np.mean(sepal_width), np.std(sepal_width)
x = np.linspace(np.min(sepal_width), np.max(sepal_width), 100)
y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / std)**2)

plt.figure(figsize=(8, 6))
plt.hist(sepal_width, bins=20, color='skyblue', edgecolor='black', alpha=0.7, density=True)
plt.plot(x, y, color='red', label='Gaussian Fit')
plt.xlabel('Sepal Width (cm)')
plt.ylabel('Density')
plt.title('Distribution of Sepal Width in Iris Dataset with Gaussian Fit')
plt.legend()
plt.show()
```

```python
# Pairplot to visualize relationships between features
sns.pairplot(IRIS, hue="species", diag_kind="kde", palette="Set1")
plt.suptitle("Feature Relationships by Species", y=1.02)
plt.show()
```

```python
# Load dataset
X = IRIS.iloc[:,:-1]
y = IRIS.iloc[:,-1]
X.head()
```

```python
# Correlation heatmap
plt.figure(figsize=(6,4))
sns.heatmap(X.corr(), annot=True)
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
X.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[0]
```

```python
# Convert back to DataFrame
scaled = pd.DataFrame(X_scaled, columns=X.columns)
scaled.head()
```

```python
fig, axes = plt.subplots(1,2, figsize=(10,4))
sns.boxplot(data=X, ax=axes[0])
axes[0].set_title("Before Scaling")

sns.boxplot(data=scaled, ax=axes[1])
axes[1].set_title("After Scaling")
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(6,4))
sns.heatmap(scaled.corr(), annot=True)
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

selector = SelectKBest(score_func=f_classif, k=2)
X_new = selector.fit_transform(X, y)

# Create dataframe of feature scores
feature_scores = pd.DataFrame({
    'Feature': X.columns,
    'F-Score': selector.scores_
}).sort_values(by='F-Score', ascending=False)

feature_scores
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=feature_scores, x='F-Score', y='Feature', palette='viridis')
plt.title("Feature Importance via ANOVA F-Test")
plt.xlabel("F-Score")
plt.ylabel("Feature Name")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_new, y, test_size=0.2, random_state=42)

print(X_train.shape, X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "SVM": SVC()
}

scores = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    scores[name] = accuracy_score(y_test, preds)

# Convert to dataframe for plotting
scores_df = pd.DataFrame(list(scores.items()), columns=["Model", "Accuracy"])
scores_df
```

```python
plt.figure(figsize=(6,4))
sns.barplot(data=scores_df, x="Model", y="Accuracy", palette="cool")
plt.title("Model Comparison (Accuracy)")
plt.ylim(0.8, 1)
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
Cancer = load_breast_cancer()
Cancer
```

```python
X, y = pd.DataFrame(Cancer.data, columns=Cancer.feature_names), pd.DataFrame(Cancer.target, columns=['typesOfCancer'])

cancer = pd.DataFrame(X, columns=Cancer.feature_names)
cancer = pd.concat([X, y], axis=1)
cancer.head()
```

```python
X.shape
```

```python
cancer.shape
```

```python
print(cancer.info())
print(cancer.describe())
print(cancer.isnull().sum())
print("Duplicates", cancer.duplicated().sum())
```

### No missing values, so no treatment
### No Duplicates, so no treatment
# Data Transformation

```python
fig = plt.figsize=(10,10)
sns.boxplot(data=cancer)
axes[0].set_title("Before Scaling")
```

```python
# scaling

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[0]
```

```python
# Convert back to DataFrame
scaled = pd.DataFrame(X_scaled, columns=X.columns)
scaled.head()
```

```python
fig = plt.figsize=(10,10)
sns.boxplot(data=scaled)
axes[0].set_title("Before Scaling")
```

```python
scaled.iloc[:, :-1]
```

```python
scaled.shape
```

```python
#train_test_split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(scaled, y, test_size=0.2, random_state=42)
print(X_train.shape, X_test.shape)
```

```python
# model training

model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python
# model testing
y_pred = model.predict(X_test)
```

```python
# Step 7: Visualize confusion matrix
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Malignant'], yticklabels=['Malignant'])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```
