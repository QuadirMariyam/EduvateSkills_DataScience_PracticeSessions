NLP is a branch of Artificial Intelligence that helps computers understand, interpret, and generate human language.
Example tasks:

Sentiment analysis (positive/negative reviews)

Chatbots (like ChatGPT!)

Translation (English → Hindi)

Text summarization

Speech recognition

# 🧹 Why Text Preprocessing?

Before feeding text into any ML or AI model, it must be cleaned and standardized because raw text is messy.
For example:

“I’m loving this product!!! 😍😍😍”
and
“I love this product.”

should both mean the same thing to your model.

# 🔧 Common Steps in Text Preprocessing

Lowercasing → Convert all text to lowercase.

Removing punctuation → Get rid of ! , ? . etc.

Tokenization → Split text into words or sentences.

Stopword removal → Remove common words like the, is, and, of.

Stemming / Lemmatization → Reduce words to their root for

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
text = "NLTK is a powerful Library for Natural Language Processing.! It helps computers understand, interpret and generate human language.!"
text
```

```python
# Step 1 - lowercase
lower_text = text.lower()
lower_text
```

```python
# Step 2 - Tokenize
sentoken_text = sent_tokenize(lower_text)
sentoken_text
```

```python
wordtoken_text = word_tokenize(lower_text)
wordtoken_text
```

```python
stop_words = set(stopwords.words('english')) # common stopword library in eng language
stop_words
```

```python
personal_stopwords = ['a', 'for', 'it', 'and']
```

```python
filtered_words = [word for word in wordtoken_text if word.isalpha() and (word not in stop_words)]
filtered_words
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]
stemmed_words
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
lemmatized_words
```

```python
# final cleaned and preprocessed text for NLP model
lemmatized_words
```

| Feature  | Stemming                | Lemmatization          |
| -------- | ----------------------- | ---------------------- |
| Output   | May not be a valid word | Always a real word     |
| Approach | Truncation rules        | Grammar + Word meaning |
| Example  | “studies” → “studi”     | “studies” → “study”    |
| Speed    | Fast                    | Slower                 |
| Use      | When speed matters      | When accuracy matters  |
